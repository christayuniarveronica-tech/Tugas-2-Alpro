```python
#Pemograman Giant

#Data diri
nama = "Giant"
usia = "20"
berat_badan = 97.5 #dalam kg
tinggi_badan = 1.75 #dalam meter
status_olahraga = False 

#Konversi dari string ke float
berat_string = 97.5
berat_float = float(berat_string)

#Kalimat pengantar
kalimat_pengantar = f"Halo, saya {nama}. Usia saya {usia} tahun dan berat saya {berat_float} kg"
print(kalimat_pengantar)
```

```python
#Menghitung BMI (Body Max Index)
bmi_asli = berat_badan / (tinggi_badan **2)
bmi_dibulatkan = round(bmi_asli, 2)

print(f"BMI Asli          : {bmi_asli}")
print(f"BMI (2 Desimal)   : {bmi_dibulatkan}")
```

```python
#Taget minum perhari

target_air = 2000  # ml
ukuran_botol = 350  # ml

# True division (desimal)
botol_true_div = target_air / ukuran_botol

# Floor division (jumlah botol penuh)
botol_penuh = target_air // ukuran_botol

# Modulo (sisa air yang belum mencapai 1 botol penuh)
sisa_air = target_air % ukuran_botol

print(f"Hasil True Division : {botol_true_div} botol")
print(f"Botol Penuh (Floor) : {botol_penuh} botol")
print(f"Sisa Air (Modulo)   : {sisa_air} ml")
```
